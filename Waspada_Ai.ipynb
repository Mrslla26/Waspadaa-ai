```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_id = "google/gemma-2-2b-it"

print("GPU tersedia:", torch.cuda.is_available())

print("Memuat tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(model_id)

print("Memuat model Gemma...")
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("Google Gemma berhasil dimuat!")
```

```python
def tanya_gemma(prompt_teks):
    messages = [
        {"role": "user", "content": prompt_teks}
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=256,
            do_sample=False
        )

    jawaban = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[-1]:],
        skip_special_tokens=True
    )

    return jawaban.strip()

print("Fungsi komunikasi Gemma berhasil dibuat!")
```

```python
import re

kata_mencurigakan = [
    "kode OTP",
    "password",
    "kata sandi",
    "transfer sekarang",
    "segera transfer",
    "klik link",
    "klik tautan",
    "hadiah",
    "pemenang",
    "rekening",
    "biaya administrasi",
    "keuntungan besar",
    "investasi",
    "data pribadi",
    "akun diblokir"
]

def deteksi_kata_mencurigakan(pesan):
    ditemukan = []

    for kata in kata_mencurigakan:
        pola = re.escape(kata)
        if re.search(pola, pesan, re.IGNORECASE):
            ditemukan.append(kata)

    return ditemukan

print("Fitur deteksi kata mencurigakan berhasil dibuat!")
```

```python
pesan_contoh = """
Selamat Anda memenangkan hadiah!
Segera klik link dan kirim kode OTP
untuk mencairkan hadiah Anda.
"""

hasil_deteksi = deteksi_kata_mencurigakan(pesan_contoh)

print("===== KATA MENCURIGAKAN =====")

if hasil_deteksi:
    for kata in hasil_deteksi:
        print("-", kata)
else:
    print("Tidak ditemukan kata mencurigakan.")
```

```python
def buat_balasan_aman(pesan):
    prompt = f"""
    Kamu adalah WASPADA AI.

    Buatkan contoh balasan yang sopan, tegas,
    dan aman untuk menghadapi pesan berikut:

    {pesan}

    Ketentuan:
    1. Jangan memberikan OTP atau kata sandi.
    2. Jangan memberikan data pribadi.
    3. Jangan menjanjikan transfer uang.
    4. Sarankan pengguna memverifikasi informasi
       melalui saluran resmi.
    5. Gunakan bahasa Indonesia yang sederhana.

    Berikan satu contoh balasan singkat.
    """

    return tanya_gemma(prompt)

print("Fitur Safe Reply Generator berhasil dibuat!")
```

```python
import sqlite3
from datetime import datetime

print("SQLite berhasil dihubungkan!")
```

```python
def buat_database():
    conn = sqlite3.connect("waspada_ai.db")
    cursor = conn.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS riwayat_analisis (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            pesan TEXT NOT NULL,
            hasil_analisis TEXT NOT NULL,
            tanggal TEXT NOT NULL
        )
    """)

    conn.commit()
    conn.close()

    print("Database WASPADA AI berhasil dibuat!")

buat_database()
```

```python
def simpan_riwayat(pesan, hasil):
    conn = sqlite3.connect("waspada_ai.db")
    cursor = conn.cursor()

    tanggal = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    cursor.execute("""
        INSERT INTO riwayat_analisis
        (pesan, hasil_analisis, tanggal)
        VALUES (?, ?, ?)
    """, (pesan, hasil, tanggal))

    conn.commit()
    conn.close()

    print("Riwayat analisis berhasil disimpan!")
```

```python
def analisis_dan_simpan(pesan):
    hasil = analisis_penipuan(pesan)

    print("===== HASIL ANALISIS WASPADA AI =====")
    print(hasil)

    simpan_riwayat(pesan, hasil)

    return hasil
```

```python
def lihat_riwayat():
    conn = sqlite3.connect("waspada_ai.db")
    cursor = conn.cursor()

    cursor.execute("""
        SELECT * FROM riwayat_analisis
        ORDER BY id DESC
    """)

    data = cursor.fetchall()
    conn.close()

    if not data:
        print("Belum ada riwayat analisis.")
        return

    for item in data:
        print("\n===== RIWAYAT ANALISIS =====")
        print("ID:", item[0])
        print("Pesan:", item[1])
        print("Hasil:", item[2])
        print("Tanggal:", item[3])
```

```python
def tampilkan_riwayat():
    conn = sqlite3.connect("waspada_ai.db")
    cursor = conn.cursor()

    cursor.execute("""
        SELECT id, pesan, hasil_analisis, tanggal
        FROM riwayat_analisis
        ORDER BY id DESC
    """)

    data = cursor.fetchall()
    conn.close()

    if not data:
        return "Belum ada riwayat analisis."

    hasil = ""

    for item in data:
        hasil += f"""
ID: {item[0]}
Tanggal: {item[3]}
Pesan: {item[1]}
Hasil Analisis:
{item[2]}

------------------------------
"""

    return hasil
```

```python
def proses_analisis(pesan):
    if not pesan.strip():
        return "Masukkan pesan terlebih dahulu.", ""

    hasil = analisis_dan_simpan(pesan)

    kata = deteksi_kata_mencurigakan(pesan)

    if kata:
        indikator = "\n".join([f"- {k}" for k in kata])
    else:
        indikator = "Tidak ditemukan kata yang cocok dengan daftar indikator."

    return hasil, indikator


def proses_balasan(pesan):
    if not pesan.strip():
        return "Masukkan pesan terlebih dahulu."

    return buat_balasan_aman(pesan)


def proses_edukasi(jenis):
    if not jenis.strip():
        return "Masukkan jenis penipuan yang ingin dipelajari."

    return edukasi_penipuan(jenis)
```

```python
with gr.Blocks(
    theme=gr.themes.Soft(),
    title="WASPADA AI"
) as app:

    gr.Markdown("""
    # 🛡️ WASPADA AI
    ### Your Smart Digital Fraud Guardian

    Asisten AI untuk membantu mengenali
    indikasi penipuan digital.
    """)

    with gr.Tab("🔍 Analisis Pesan"):
        gr.Markdown("Masukkan pesan yang ingin diperiksa.")

        input_pesan = gr.Textbox(
            label="Pesan yang Diterima",
            placeholder="Tempel pesan WhatsApp, SMS, atau DM di sini...",
            lines=6
        )

        tombol_analisis = gr.Button(
            "Analisis Sekarang",
            variant="primary"
        )

        output_analisis = gr.Textbox(
            label="Hasil Analisis WASPADA AI",
            lines=10
        )

        output_indikator = gr.Textbox(
            label="Kata atau Frasa Mencurigakan",
            lines=4
        )

        tombol_analisis.click(
            fn=proses_analisis,
            inputs=input_pesan,
            outputs=[output_analisis, output_indikator]
        )

    with gr.Tab("💬 Safe Reply"):
        gr.Markdown("""
        Buat balasan yang sopan dan aman
        untuk menghadapi pesan mencurigakan.
        """)

        input_balasan = gr.Textbox(
            label="Pesan yang Diterima",
            lines=5
        )

        tombol_balasan = gr.Button("Buat Balasan Aman")

        output_balasan = gr.Textbox(
            label="Saran Balasan",
            lines=6
        )

        tombol_balasan.click(
            fn=proses_balasan,
            inputs=input_balasan,
            outputs=output_balasan
        )

    with gr.Tab("📚 Scam Knowledge"):
        gr.Markdown("""
        Pelajari berbagai jenis penipuan digital
        dan cara menghindarinya.
        """)

        input_edukasi = gr.Textbox(
            label="Jenis Penipuan",
            placeholder="Contoh: Phishing, penipuan hadiah..."
        )

        tombol_edukasi = gr.Button("Pelajari")

        output_edukasi = gr.Textbox(
            label="Informasi Edukasi",
            lines=10
        )

        tombol_edukasi.click(
            fn=proses_edukasi,
            inputs=input_edukasi,
            outputs=output_edukasi
        )

    with gr.Tab("📋 Riwayat Analisis"):
        gr.Markdown("Lihat hasil pemeriksaan sebelumnya.")

        tombol_riwayat = gr.Button("Tampilkan Riwayat")

        output_riwayat = gr.Textbox(
            label="Riwayat Pemeriksaan",
            lines=15
        )

        tombol_riwayat.click(
            fn=tampilkan_riwayat,
            inputs=None,
            outputs=output_riwayat
        )

    gr.Markdown("""
    ---
    **WASPADA AI** | Powered by Google Gemma 2

    Hasil analisis merupakan indikasi risiko,
    bukan kepastian bahwa suatu pesan adalah penipuan.
    """)

print("Tampilan WASPADA AI berhasil dibuat!")
```
